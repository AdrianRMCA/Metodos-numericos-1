# Método de Newton

 ### Ejemplo 1
 #### Considere la función f(x) = cos(x) - x = 0.
 #### Aproxime una raíz de f usando el método de Newton.


El objetivo es encontrar una solución aproximada de la ecuación
$\cos(x)-x=0$, considerando el intervalo $[0,1]$.

Para utilizar el método de Newton-Raphson, primero establecemos
la función original y calculamos su primera derivada:

$$
f(x)=\cos(x)-x
$$

$$
f'(x)=-\sin(x)-1
$$

Este método se fundamenta en la aproximación lineal obtenida
mediante el desarrollo de Taylor. A partir de un valor inicial
$p_0$, se generan aproximaciones sucesivas que permiten
acercarnos progresivamente a la raíz de la ecuación.

Para obtener cada nueva aproximación, se emplea la siguiente
fórmula iterativa:

$$
p_n=p_{n-1}-\frac{f(p_{n-1})}{f'(p_{n-1})}
$$

Sustituyendo la función y su derivada, obtenemos:

$$
p_n=p_{n-1}-\frac{\cos(p_{n-1})-p_{n-1}}{-\sin(p_{n-1})-1}
$$

De esta manera, partiendo de $p_0=0.5$, aplicaremos
10 iteraciones para aproximar la raíz de la ecuación
y analizar la disminución del error entre los resultados.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Definimos la función original
def f(x):
    return np.cos(x) - x

# Evaluamos la función en 0.5
f(0.5)
```

```python
# Generamos 100 valores entre 0 y 1
xx = np.linspace(0, 1, 100)

# Graficamos la función
plt.plot(xx, f(xx))

# Dibujamos la línea horizontal en cero
plt.axhline(y=0, color="red")

# Agregamos título y nombres a los ejes
plt.title("Gráfica de f(x) = cos(x) - x")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```

```python
# Definimos la derivada de la función
def df(x):
    return -np.sin(x) - 1

# Evaluamos la derivada en 0.5
df(0.5)
```

```python
# Definimos el método de Newton
def newton(p0, n):

    lista = []   # Guarda las aproximaciones
    error = []   # Guarda los errores

    # Realizamos las iteraciones
    for i in range(n):

        fp0 = f(p0)   # Evaluamos la función
        dif = df(p0)  # Evaluamos la derivada

        # Calculamos la nueva aproximación
        p = p0 - (fp0 / dif)

        # Guardamos la aproximación
        lista.append(p)

        # Calculamos el error absoluto
        e = np.abs(p - p0)
        error.append(e)

        # Actualizamos el valor inicial
        p0 = p

    return lista, error
```

```python
# Ejecutamos Newton con 10 iteraciones
p = newton(0.5, 10)

# Mostramos las aproximaciones y errores
pd.DataFrame(p, index=["Aproximación", "Error"]).T
```
