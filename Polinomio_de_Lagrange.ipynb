```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
 return np.cos(x)
```

```python
def p(x):
  return -0.432*(x**2)-0.032*x+1
```

```python
x=np.linspace(0,1)
plt.plot(x,f(x))
plt.plot(x,p(x))
```
